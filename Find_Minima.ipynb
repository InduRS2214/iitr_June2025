```python
# Convex Function

def find_minima_func(f, lower = -10, upper = 10, tol = 1e-7):
    x_min = 0

    while upper - lower > tol:
        mid1 = lower + (upper - lower) / 3
        mid2 = upper - (upper - lower) / 3
        f1 = f(mid1)
        f2 = f(mid2)
        if f1 < f2:
            x_min = mid1
            upper = mid2
        else:
            x_min = mid2
            lower = mid1
    return x_min, f(x_min)

```

```python
def test_function(x):
    return (x - 3)**2 + 5

find_minima_func(test_function)
```

```python
import math
# ------------------------
# Test Cases
# Each test: (function, expected_x, expected_f)
tests = [
    # Simple quadratic
    (lambda x: (x - 2)**2, 2, 0),

    # Shifted quadratic
    (lambda x: (x + 3)**2 + 1, -3, 1),

    # Scaled quadratic
    (lambda x: 5*(x - 1)**2 + 10, 1, 10),

    # Quartic convex
    (lambda x: x**4 + 2*x**2 + 3, 0, 3),

    # Absolute value (nondifferentiable)
    (lambda x: abs(x - 5), 5, 0),
    
        # Monotonic convex function (minimum at lower bound)
    (lambda x: math.exp(x), -10, math.exp(-10)),

    # Exponential + quadratic (approximate)
    (lambda x: math.exp(x) + (x - 1)**2, 0.44285, None),


]

# ------------------------
# Run Tests
for i, (f, expected_x, expected_f) in enumerate(tests, 1):
    x_min, f_min = find_minima_func(f)
    print(f"Test {i}: x_min = {x_min:.6f}, f_min = {f_min:.6f}")
    
    # Check if x_min is within tolerance
    assert abs(x_min - expected_x) < 1e-2, f"Test {i} failed: x mismatch"
    
    # Check f_min if exact value is known
    if expected_f is not None:
        assert abs(f_min - expected_f) < 1e-2, f"Test {i} failed: f(x) mismatch"

print("All tests passed!")
```
