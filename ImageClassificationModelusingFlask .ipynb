virtual env -

Flask - framework for web development


1. Create Project directory 'Image-Classification-App'
2. Create a virtual environment 'Img-Class-Env' --> python3 -m venv Img-Class-Env
   Activate virtual environment --> source Img-Class-Env/bin/activate
3. Installing necessary packages
     - package manager --> pip install --upgrade pip
     - Flask package --> pip install Flask==1.1.2
     - tensorflow package --> { python -m pip install --upgrade setuptools
                                pip install protobuf==3.17.3
                                pip install --no-cache-dir  --force-reinstall -I grpcio==1.11.0
                                pip install tensorflow==1.14.0 }
     - pillow package --> pip install pillow==6.2.2
4. Create Helper Function 'app_helper.py'
     - import the pre-trained model
     - preprocess the input image as per requirements of the pre-trained model
     - feed this preprocessed image as the input of the model and get the top 3 predictions as classified by the model.
5. Create Static Directory 'Image-Classification-App/static'
     - contain static files such as JS, CSS, and images by Flask.
6. Create CSS Directory & css file 'Image-Classification-App/static/css/main.css'
     -  contains the formatting of the HTML files that we will be creating
7. Create Uploads directory 'Image-Classification-App/static/uploads'
     - store the image uploaded to our app. This stored image will be used to display it along with the predictions.
8. Create Template directory 'Image-Classification-App/templates'
     - contain the HTML files by Flask
9. Create layout.html 'Image-Classification-App/templates/layout.html'
     - defines the layout of the pages, like the Navigation bar, which should be the same for the pages.
     - Python-Flask code snippets into an HTML file using {{ }}
     - linking the CSS file Image-Classification-App/static/css/main.css
10. Create index.html 'Image-Classification-App/templates/index.html'
     - home page, where the user is asked to upload the image whose class should be predicted using our pre-trained model.
     - imported layout.html
     - form for uploading an image and submit it.
11. Create upload.html 'Image-Classification-App/templates/upload.html'
     - displays the uploaded image along with the predicted classes and the probabilities.
     - import/link the layout.html
     - page content 
            - display input image we uploaded along with top 3 predictions given by the classifier for this input image
            - form to upload and submit another input image
12. Create app.py 'Image-Classification-App/app.py'
     -  root of our Flask application which we will run in the command line prompt.
     -  we define all the routes and functions to perform for each action. 
13. Run app.py --> command <<python app.py>> and then open "http://e.cloudxlab.com:4100/" in Google chrome
     
     
Remove Virtual Memory as it consumes more memory.

        - Commands: { deactivate, 
                      cd ~
                      cd Image-Classification-App
                      rm -r Img-Class-Env   }
            

In the future, when we want to run this web app again, we just need to 

 - (1)switch to the directory Image-Classification-App and create the virtual environment again in this project directory, 
 - (2) activate the virtual environment 
 - (3) install the necessary packages 
 - (4) and run the app, as mentioned in the corresponding slides.
