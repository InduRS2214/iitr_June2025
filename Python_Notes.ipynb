```python
#Lambda Function

def myfunc(n):
    return lambda x: x*n

double = myfunc(5)
print(double(2))

triple = myfunc(6)
print(triple(3))

print(myfunc(5)(2))
```

```python
#combined for loop and return statement

foo = [2, 18, 9, 22, 17, 24, 8, 12, 27]

def filter_divisbile_by_n(n, numlist):
    return [num for num in numlist if num%n==0] #####

divisible_by_3 = filter_divisbile_by_n(3, foo)

print("Method 1 output is:")
print(divisible_by_3, "\n")
```
