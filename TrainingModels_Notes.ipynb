Linear Regression
- Linear Regression is a supervised learning algorithm used to predict continuous values by fitting the best straight line through the data. Formula: y=wX+b, Where: w = weight (slope) & b = bias (intercept)
- 2 Approaches for training linear regression model
    - 1. Normal Closed-form Equation
    - 2. Gradient Descent

#5
Types of GD:
1. Batch GD
- Uses the entire dataset for each update.
- calculate MSE on entire datset and tweak theta based on the gradients. This happens in every epoch
- formula: theta_new = theta - eta * gradients, gradients is the rate of change of error w.r.t theta (partial derivative)
- 
- Pros:
    - stable
    - smooth convergence
- Cons:
    - slow for large datasets
    - will stuck in local minima
2. Stochastic GD (SGDRegressor)
- Uses only one random training example at a time.
- Pros:
    - faster
    - can escape shallow regions faster
    - will not stuck in local minima
- Cons:
    - noisy updates
    - fluctuates around minimum
- SGD is useful for large datasets.
- Simulated Annealing: Start with larger steps that helps quick progress and escape local minima and then steps get smaller and smaller which helps alogorithm to settle in global minimum. Resembles the process of annealing in metallurgy.
- Learning Schedule: Function that determines learning rate at each iteration
3. Mini-batch GD
- Instead of computing gradients on whole dataset (Batch GD) or just one random instance (SGD), mini-batch GD computes gradients on small random dataset (like 32, 64, 128 samples).
- Combines Batch + SGD.
- Pros:
    - faster than batch
    - less noisy than SGD
    - most commonly used
- will stuck in local minima
- Best practical choice.

#6
Learning Rate: 
- step size used to update the model parameters in each iteration. (How big a step should we take toward the minimum?)
- Small learning rate → tiny steps → safe but slow
- Large learning rate → big jumps → fast but may overshoot
- Proper learning rate → reaches minimum efficiently
