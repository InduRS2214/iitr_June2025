```python
import os
def file_read_func(filepath):
    if not(os.path.exists(filepath)):
        return -1
    else:
        with open(filepath) as f:
            
            a = []
            for line in f:
                for i in line:
                    if i == " " or i == '\n':
                        continue
                    else:
                        a.append(i)
                print(a)
                
        return len(a)
file_read_func("/cxldata/python_sample_file")        
```

```python
def dict_func(dic):
    words = dic.split()
    new_dict = {}
    for i in words:
        count = 0
        for j in words:
            if(i==j):
                count = count + 1
        new_dict[i] = count
    return new_dict
dict_func('the quick brown fox jumps over the lazy dog')   
```

```python
import socket
import sys

hostname="www.bing.com"

try:
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    print("Socket successfully created here")
except socket.error as err:
    print("socket creation failed here with error %s" %(err))
    
port=80 # default port for socket

try:
    host_ip = socket.gethostbyname(hostname)
except socket.gaierror:
    print("Error resolving the host")
    sys.exit()
    
s.connect((host_ip, port)) #connecting to server

print(f"Socket successfully connected to {hostname}")
```
