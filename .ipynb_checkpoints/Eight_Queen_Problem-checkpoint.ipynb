```python
def create_board():
    board = []
    for i in range(8):
        row = []
        for j in range(8):
            row.append(False)
        board.append(row)
    #print(board)
    return board

def is_safe(board, row, column):
    for i in range(row):
        if board[i][column]:
            return False
    
    i, j = row, column
    while i >= 0 and j >= 0:
        #print("checking diagonal")
        if board[i][j] == True:
            return False
        i -= 1
        j -= 1
        
    i, j = row, column
    while i >= 0 and j < 8:
        #print("checking anti-diagonal")
        if board[i][j] :
            return False
        i -= 1
        j += 1  
        
    return True

def find_positions(board, row):
    positons = []
    for j in range(8):
        #print(board[row][j])
        if board[row][j] == False and is_safe(board,row,j):
            positons.append(j)
    return positons

def print_board(board):
    for i in range(8):
        for j in range(8):
            if board[i][j]:
                val = 'Q'
            else: val = '.'
            print(f"{val} ", end='')
        print('\n')
        
# code from chatGPT for printing
#def print_board(board):
#   for row in board:
#        print(" ".join("Q" if cell else "." for cell in row))
#    print()


def solve(board, queens = 0):
    if queens == 8:                    # base case: all queens placed
        print_board(board)
        print("=============================================================================")
        return True                       # stop after first solution
    
    count = 0
    row = queens
    positions = find_positions(board, row)
    for position in positions:
        board[row][position] = True
        count += solve(board, queens = queens + 1)
        board[row][position] = False       # back tracking
    
    return count                       # count number of solutions

board = create_board()
solve(board)
```
