```python
import pandas as pd

pd_obj = pd.Series([-2,1,5,4]) # Series is a column in excel sheet
print(pd_obj)  # will print <name> <coulumn_values>


```

```python
s = pd.Series([-2,1,5,4], index=["alice", "henk", "joel", "Holly"]) # <named index> 
#index has to be unique(looks like a hash table)
print(s)

print('\n')

#default values & named series object
a = pd.Series(42,index=["abc","bcd","cde","def"], name="Named_SeriesObj")
print(a)
```

```python
#print(s[1])
print(s.iloc[1]) # accessing by position
print(s.loc["Holly"]) # accessing by index name
```

```python
weights = {"x":32,"y":54,"z":43, "a":67}
s1 = pd.Series(weights) # Series objects can create from dictionary
print(s1)
s2 = pd.Series(weights, index=["x","a"])
print(s2)
```

```python
# when s1 + s2, eventhough those are int64, sum result will be float64 by default to include all data types
# if there is no matching index in both, sum result will have unmatched indexes with values as "NAN" (Not a Number) 
# and matching indexes values will sum up
```

```python
print(pd.DataFrame([s,s1,s2])) # indexes will be the column names in dataframe
#when series objects are having different indexes, those many columns will be created in dataframe
```

```python
import pandas as pd
import matplotlib.pyplot as plt

marks = {'m1':pd.Series([24,25,28], index=["Maths", "Science", "English"]), 'm2':pd.Series([14,20,22], index=["Maths", "Science", "English"]), 'm3':pd.Series([2,5,8], index=["Maths", "Science", "English"])}
print(pd.DataFrame(marks))
```

```python
#marks = {'m1':pd.Series([24,25,28], index=["Maths", "Science", "English"])}
df = pd.DataFrame(marks)
print(df)
print('\n')
print(df['m2']) # 2nd column, m2
print('\n')
print(df.iloc[0]) #0th raw, Maths
print('\n')
print(df.loc["English"]) # 3rd row, English
```

```python
import pandas as pd
import matplotlib.pyplot as plt

marks_values = {'maths':pd.Series([24,25,28], index=["Alice", "James", "Holly"]), 'Science':pd.Series([14,20,22], index=["Alice", "James", "Holly"]), 'English':pd.Series([2,5,8], index=["Alice", "James", "Holly"])}
marks = pd.DataFrame(marks_values) 
print(marks)
```

```python
#Adding new column 'Total' by summing up all subjects marks and adding new column 'square' by squaring 'Total' values using 'assign' and a function 'myfunc'
marks_values = {'maths':pd.Series([24,25,28], index=["Alice", "James", "Holly"]), 'Science':pd.Series([14,20,22], index=["Alice", "James", "Holly"]), 'English':pd.Series([2,5,8], index=["Alice", "James", "Holly"])}
marks = pd.DataFrame(marks_values) 
marks["Total"] = marks.sum(axis=1)
def myfunc(x):
    return x["Total"]**2
marks = marks.assign(square = myfunc) # without assigning to a variable assign methos will not changing existing dataframe 
print('\n')
print(marks)
```

```python
#Querying dataframe

marks.query("Total>=50 and English>5")
```

```python
# by default dataframe sorts in index ascending order
marks.sort_index(ascending=False)

marks.sort_index(inplace = True, ascending=False) # inplace=True will update dataframe in memory

#sort by value
marks.sort_values(by="Total")
```

```python
#plot dataframe

marks.plot(kind="line",  y=["maths", "Science","English"])
plt.show()
```
