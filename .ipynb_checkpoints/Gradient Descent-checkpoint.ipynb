```python
#closure in python

def Ec(c):
    def cal(m):
        return E(m,c)
    return cal

ec = Ec(2)
ec(1)
```

```python
#sample code

def E(m,c):
    return (m*1 + c - 8)**2 + (m*2 + c - 11)**2 + (m*3 + c - 16)**2

def diffp(fn, idx, *args):
    eta = 0.000001
    args = list(args)
    y = fn(*args)
    args[idx] += eta
    y1 = fn(*args)
    return (y1-y)/eta



epoc = 100
m, c = np.random.random(2,)
for i in range(epoc):
    eta = 0.01
    gradm = diffp(E, 0, m, c)
    gradc = diffp(E, 1, m, c)
    m = m - gradm * eta
    c = c - gradc * eta
    print(m, c, E(m, c))
```

```python
#### Model Creation

import numpy as np


# Data Generation
#---------------------------------
x1 = np.random.random(400)
x2 = np.random.random(400)
noise = (np.random.random(400) - 0.5)/4
y = 3*x1 + 5*x2 + 10 + noise


# Error Function
# ---------------------------------------
def error(m1, m2, c):
    pred = m1*x1 + m2*x2 + c
    return np.mean((pred - y)**2)


# Find gradient
#--------------------------------------

def diffp(fn, idx, *args):
    delta = 0.000001
    args = list(args)
    y = fn(*args)
    args[idx] += delta
    y1 = fn(*args)
    return (y1-y)/delta


# Model
# -----------------------------------------------
m1, m2, c = np.random.random(3,)
```

```python
#### Training part (Adjust weights value to minimize error)

epoc = 100 
for i in range(epoc):
    eta = 0.19
    gradm1 = diffp(error, 0, m1, m2, c)
    gradm2 = diffp(error, 1, m1, m2, c)
    gradc = diffp(error, 2, m1, m2, c)
    m1 = m1 - gradm1 * eta
    m2 = m2 - gradm2 * eta
    c = c - gradc * eta
    print(i, m1, m2, c, error(m1, m2, c))
```

```python
x1 = np.arange(400)
x2 = np.arange(400)
noise = np.random.random()
y = 3*x1 + 5*x2 + 10 + noise
#print(x1, x2, y)

def error(m1, m2, c):
    pred = m1*x1 + m2*x2 + c
    return np.sum((pred - y)**2)
```
