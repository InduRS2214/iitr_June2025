Random Forest
- ensemble of decision tree
- trained via bagging method
- max_samples=size of training data
- majority voting

- RandomForestClassifier & RandomForestRegressor

- introduces extra randomness when growing trees
- instead of searching for the very best feature when splitting a node, it searches for the best feature among a random subset of features
- this results in greated tree diversity, which trades higher bias for a lower variance => yielding an overall better model

Extremly Randomized Trees Ensemble (Extra Trees)
- it is possible to make trees even more random by also using random thresholds for each feature rather than searching for the best possible thresholds, like regular decision trees do
- forest of such extremely random trees => Extra Trees
- Extra-Trees much faster to train than regular Random Forests since finding the best possible threshold for each feature at every node takes more time

Relation of Bias & Variance on underfitting and overfitting
- Bias => error due to overly simple assumptions
- High Boas => Model is too simple to capture the patterns ====>> UnderFitting
- ==> causes poor train & test performance
- Variance => error due to too much sensitivity on training data
- High Variance => Model learns noise instead of actual pattern ====>> OverFitting
- ==> causes very good train peroformance but poor test performance

Boosting (hypothesis boosting)
- any ensemble methods that combines several weak models into a stronger one by training predictors sequentially each trying to corrects its predecessor
    - AdaBoost (Adaptive Boosting)
        - similar to Gradient Descent
        - Instead of tweaking a predictor's parameters to minimize a cost function, adaboost adds predictors to ensemble, gradually making it better
        - predictors will have diffrent weights
        - predecessor will boost the weights of misclassified instances by former one
        - adaboost will theoretically use many algorithms, but generally use same alogorithms 
        - Base estimator could be anything, decision tree, SVM etc
        - cannot scale/bagging/pasting since it train models/algorithms sequentially
        - The more accurate the predictor, the higher is the weight will be.
        - if it is just guessing randomly, then its weight will be close to zero
        - Algorithms/formulas
            - R = sum of weights of misclassified instances/sum of weights of total instances
            - predictor weight, Alpha = eta * log ((1-R)/R)
            - instance weights, w(i) = w(i) for right classsified instances   or   w(i)*exp(alpha(i)) for misclassified instance
            - To make predictions,
                - compute predictions of all predictors
                - weighs them using predictor weight, alpha
                - predicted class is the one with majority of weighted votes(soft)        
         - SAMME16 -> multiclass version of Adaboost -> 2 versions => real value version & Integer value version
    - Gradient Boosting 
        - similar to adaboost, works by sequentially adding predictors to an ensemble, each one correcting its predecessor
        - But instead of tweaking instance weights in each iteration, this method tries to fit new predictor to the residual errors made by previous predictor
        - Prediction on new instance will adding up the predictions of all predictors in ensemble
        - GradientBoostingRegressor (GBRT)
        - if learning rate is less, more number of trees would be needed for fitting the training dataset
        - reduce overfit:
            - reduce number of estimators or
            - flatten/reduce depth of decision trees
        - Early stopping by warm_start=True => In Gradient Boosting, warm_start=True allows the model to retain previously trained trees and continue adding new trees instead of retraining from scratch, making iterative tuning and early stopping more efficient
        - Stochastic Gradient Boosting - use subsample parameter to specify each tree should trained with given % of training dataset. This technique again will speed up training with higher bias & lower variance.
        - GB Stacking - is an ensemble method - instead of using functions to aggregate predictions of all predictors, train a model for doing this aggregation. [for eg, there is 2 subset of data and 3 base models,Then in stacking 1st layer will use subset1 for training and do predictions on second set and these predictions will be training set of 2nd layer and blender will use the last layer predictions as training set]
        - Blending - merging of predictions of all predictors using another algorithm
- Nowadays adaboost and Gradientboosting almost replaced by deep learning
- Learning rate -> in what ratio boost weights should updated

Regularization
- to avoid overfitting by adding penalty with cose function. penalty punishes model complexity
- L1 Regularization (Lasso)
    - cost = error + lambda * |w|   , w-> weight
- L2 Regularization (Ridge)
    - cost = error + lambda * w^2   
- lambda controls penlty strength
    - 0 => no regularization, model overfit
    - small value => slight control on model complexity
    - large value => strong simplification, model underfit
